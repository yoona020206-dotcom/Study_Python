# 클로저(closure)
- 함수 안에 함수를 선언 하는 함수

```python
def func_out(out_param):
    a = 10
    print(f"a:{a} & out_param: {out_param}")
    return a + out_param

func_out(5)
```

```python
def func_out(out_param):
    a = 10
    print(f"a:{a} & out_param: {out_param}")

    def func_in():
        return a + out_param  # 함수도 데이터가 가능 

    return func_in

func_out(5)
```

```python
def func_out(out_param):
    a = 10
    print(f"a:{a} & out_param: {out_param}")

    def func_in():
        return a + out_param  # 함수도 데이터가 가능 

    return func_in

func_out(5)() #함수 실행은 () 
```

```python
def func_out(out_param):
    a = 10
    #print(f"a:{a} & out_param: {out_param}")

    def func_in(qwer):
        return a + out_param +qwer # 함수도 데이터가 가능 

    return func_in

func_out(5)(3) #함수 실행은 () 
```

# 콜백 함수
- 함수를 입력변수로 실행하는 함수

```python
def add_one(num):
    return num +1

add_one(4)
```

```python
def add_three(num):
    return num +3

add_three(4)
```

```python
def call_func(param:int, fucn1, fucn2):
    if param %2 == 0:
        return fucn1(param) # add-one(4)

    return fucn2(param) # add-three(5)> 5+3

call_func(4, add_one, add_three)
```

```python
def call_func(param:int, fucn1, fucn2):
    if param %2 == 0:
        return fucn1(param) # add-one(4)

    return fucn2(param) # add-three(5)> 5+3

call_func(5, add_one, add_three)
```

```python
def call_func(param:int, *args_funcs):
    if param % 2 == 0:
        return args_funcs[0](param) # add_one(4)
    
    return args_funcs[1](param) # add_three(5) > 5 + 3

call_func(8, add_one, add_three)
```

# Decorate
- 클로저 함수: 함수 안에 함수를 호출하는 경우
- 콜백 함수: 입력 변수에 함수를 적용한 경우

```python
def decotator_func(org_func): #함수를 인자로 받는다
    def wrapper_func(): # 내부 함수를 구현
        print("org_func가 실행되기 전 입니다.")
        org_func() # org_func 함수를 실행
    return wrapper_func # 내부함수의 주소를 반환
```

```python
@decotator_func # Decorate 함수 사용
def do_func():
    print("original함수가 실행 되었습니다.")

do_func() # 진짜 실행된 함수는 warpper_func()
```
