- 예시 >> [8, 6, 2, 8, 5, 1, 2, 5, 4, 9]
        - 최소값: 1 / 최대값: 9 / 중간값: 5

```python
# 1.리스트에 있는 데이터 정렬 => sort()
# 2.가장 작은 수 찾기 [0]
# 3. 가장 큰 수 찾기 [-1]
# 4. 중간 값/ 중위값 
# 4-1. 만약 전체 데이터의 수가 홀 수 이면, 가운데 값 찾기 => %2 != 0
# 4-2. 만약 전체 데이터의 수가 짝 수 이면, 가운데의 값의 평균값 사용하기  
# => len()/2 

i = [8, 6, 2, 8, 5, 1, 2, 5, 4, 9]

i .sort() # 정리

print("최대 값:",i[0], "최소 값:", i[-1]) 
l = len(i)
if l%2 == 0:
    mid = (i[(l//2)-1] + i[(l//2)])/2
else:
    mid = i[(l//2)]

print("중간 값", mid)
```

```python
# 1.리스트에 있는 데이터 정렬 => sort()
lst = [8, 6, 2, 8, 5, 1, 2, 5, 4, 9]

lst.sort() #정렬

lst #확인
```

```python
# 2.가장 작은 수 찾기 [0]

minium = lst[0]
minium
```

```python
# 3. 가장 큰 수 찾기 [-1]

maximum= lst[-1]
minium
```

```python
# 4. 중간 값/ 중위값 
if len(lst) %2 != 0:
    # 4-1. 만약 전체 데이터의 수가 홀 수 이면, 가운데 값 찾기 => %2 != 0
    print("홀수")
    median = lst[len(lst)//2]
else:
    # 4-2. 만약 전체 데이터의 수가 짝 수 이면, 가운데의 값의 평균값 사용하기  
    # => len()/2 
    print("짝수")
    median = lst[len(lst)//2-1] + lst[len(lst)//2]/2
```

```python

print(f"최소값:{minium}, 최대값: {minium}, 중위값: {int(median)}")
```
