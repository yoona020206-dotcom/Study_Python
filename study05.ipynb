# (공통) 튜플 & 리스트
- 여러개의 데이터를 저장하는 것
### 리스트 vs 튜플
- 리스트
    - 생성, 조회, 수정 등
- 튜플
    -생성, 조회만 가능> 수정 불가!

```python
# 튜플 상수처럼
# 리스트 변수처럼
# 리스트 []
# 튜플은 ()
a = (1, 2, 3)
a
```

```python
type(a)
```

```python
len(a)
```

```python
a[0] = 10 # 안에 있는 데이터 수정 불가 하지만 a = 10은 가능
```

# (공통) 딕셔너리
- 여러 데이터를 저장하는 것

### 딕셔너리 vs그외 (리스트, 튜플)
- 리스트 [1,  2, 3]
    - 생성, 조회, 수정 등 
    - 조회: 인덱스(정수)를 이용
- 튜플 (1, 2, 3)
    - 생성, 조회, 수정 불가
    - 조회: 인덱스(정수)를 이용
- 딕셔너리 (키 & 벨류 쌍) {"키1": "벨류1", "키2": "벨류2"}
    - 생성, 조회, 수정 등
    - 조회: 인덱스가 아닌 키를 이용

```python
dic = {"one": 1, "two": 2, "three": 3}
lst = [1, 2, 3]
tup = (1, 2, 3)
```

```python
lst[0] =10
# tup[0] =10 오류 발생!!
dic["one"] = 10
```

```python
lst, dic
```

```python
# [25,300,5000]
{
    "age":25,
    "height":300,
    "weight":5000
}
```
