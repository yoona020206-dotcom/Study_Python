# 클래스

```python
# 클래스 선언
class GrandMother:
    # 변수 선언
    family = "grandparents"

    # 함수 선언
    def print_self(self): # ()안에 입력 안하면 오류
        print(self)
```

```python
# 인스턴스화: 클래스를 메모리에 올리는 행위
# -> 클래스가 메모리에 올라간 것을 "인스턴스/객체"라고 함 

# grandMother() > 인스턴스화
# LEE > 인스턴트/ 객체
LEE = GrandMother()
```

```python
type(GrandMother)
```

```python
type(LEE)
```

```python
LEE.family # 객체(클래스)의 변수 호출
```

```python
LEE.print_self() # 객체(클래스)의 함수 호출, ()안에 self가 들어감, (self란 인스턴스/객체입니다.)
```

```python
GrandMother.print_self() #입력 변수 self 들어가지 않은 상태에서 함수 호출 -> 오류가 발생  grandMother는 클래스, 즉 인스턴스가 포함 되지 않음
```

```python
# 클래스 선언
class grandMother:
    # 변수 선언
    family = "grandparents"

    # 함수 선언
    def print_self(self): # ()안에 입력 안하면 오류
        print(self)

    def print_hello(self, p_name:str= "홍길동")-> None: 
        print(f"{p_name}님, 안녕하세요.")  # 클래스를 다시만들면 인스턴스도 다시 만들기
```

```python
# 클래스 선언
class grandMother:
    # 변수 선언
    family = "grandparents"

    # 인스턴트화를 실행할 함수 선언, 매직 매쏘드 시작과 끝에 __ 붙이기
    # __init__ 생성함수 
    # => 메소드에서 사용할 변수들을 정의/선언/초기화
    def __init__(self, p_name: str, p_age: int)-> None:
        print("생성함수 실행 했어요~")
        self.name = p_name if p_name is not None else "없음"
        self.age = p_age if p_age is not None and isinstance(p_age, int) else 0

    def print_hello(self)-> None: 
        print(f"{self.name}님({self.age}), 안녕하세요.")  # 클래스에서 가져온 변수 씀
```

```python
LEE = grandMother("홍길동", 33)
```

```python
LEE.name
```

```python
LEE.age
```

```python
LEE.print_hello() # 객체에서 선언한 변수는 수정 호출 가능
```

```python
# 클래스 선언
class grandMother:
    # 변수 선언
    family = "grandparents"

    # 인스턴트화를 실행할 함수 선언, 매직 매쏘드 시작과 끝에 __ 붙이기
    # __init__ 생성함수 
    # => 메소드에서 사용할 변수들을 정의/선언/초기화
    def __init__(self, p_name: str, p_age: int)-> None:
        print("생성함수 실행 했어요~")
        self.name = p_name if p_name is not None else "없음"
        self.age = p_age if p_age is not None and isinstance(p_age, int) else 0

    def print_hello(self, p_new_name:str=None)-> None: 
        if p_new_name is not None:
            #데이터가 존재 한다면,
            self.name = p_new_name #수정
        print(f"{self.name}님({self.age}), 안녕하세요.")  # 클래스에서 가져온 변수 씀
```

```python
LEE = grandMother("홍길동", 33)
```

```python
LEE.print_hello("신사임당")
```
