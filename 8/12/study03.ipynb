```python
# 클래스 선언
class GrandMother:
    # 변수 선언
    family = "grandparents"

    # 인스턴트화를 실행할 함수 선언, 매직 매쏘드 시작과 끝에 __ 붙이기
    # __init__ 생성함수 
    # => 메소드에서 사용할 변수들을 정의/선언/초기화
    def __init__(self, p_name: str, p_age: int)-> None:
        print("생성함수 실행 했어요~")
        self.name = p_name if p_name is not None else "없음"
        self.age = p_age if p_age is not None and isinstance(p_age, int) else 0 # isinstance(p_age, int) 는 함수 is로 시작하면 무조건 bool값만 나옴

    # 함수 선언
    def print_hello(self, p_new_name:str=None)-> None: 
        if p_new_name is not None:
            #데이터가 존재 한다면,
            self.name = p_new_name #수정
        print(f"{self.name}님({self.age}), 안녕하세요.") 
```

```python
class Father(GrandMother):

    def print_goodjob(self):
        print("good job")
```

```python
LEE = Father(p_name= "홍길동", p_age = 32)
KIM = GrandMother("신사임당",18)
```

```python
LEE.print_hello()
```

```python
LEE.print_goodjob()
```

```python
KIM.print_hello()
```

```python
KIM.print_goodjob() # 할머니는 아빠의 코드를 가지고있지 않음
```
