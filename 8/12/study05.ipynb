```python
class MyDataSet:

    def __init__(self, data):
        print("생성함수야~~~")
        self.data = data

    def __call__(self, param):  #객체를 함수처럼 실행 
        print(f"{param} 데이터가 들어왔어요~")
        return len(self.data)
```

```python
dt = MyDataSet(list(range(50,100)))
```

```python
dt("홍길동") #객체를 함수처럼 실행 
```

```python
class MyDataSet:

    def __init__(self, data: list) -> None:
        print("생성함수야~~~")
        self.data = data

    def __call__(self, param: str) -> int:  #객체를 함수처럼 실행 
        print(f"{param} 데이터가 들어왔어요~")
        return len(self.data)

    def __getitem__(self, idx: int): # 요소를 호출 할때 사용
        print(f"{idx}가 들어갔어요~~~")
        return self.data[idx]
```

```python
dt = MyDataSet(list(range(50,100)))
```

```python
dt("홍길동")
```

```python
dt[-1] # 마지막 숫자는 포함 안됨
```

```python
class MyDataSet:

    def __init__(self, data: list) -> None:
        print("생성함수야~~~")
        self.data = data

    def __call__(self, param: str) -> int:  #객체를 함수처럼 실행 
        print(f"{param} 데이터가 들어왔어요~")
        return len(self.data) * 10

    def __getitem__(self, idx: int): # 요소를 호출 할때 사용
        print(f"{idx}가 들어갔어요~~~")
        return self.data[idx]

    def __len__(self):
        print("코딩 좋아 ㅎㅎ")
        return len(self.data)
```

```python
dt = MyDataSet(list(range(50,100)))
```

```python
class MyDataSet:

    def __init__(self, data: list) -> None:
        print("생성함수야~~~")
        self.data = data

    def __call__(self, param: str) -> int:  #객체를 함수처럼 실행 
        print(f"{param} 데이터가 들어왔어요~")
        return len(self.data) * 10

    def __getitem__(self, idx: int): # 요소를 호출 할때 사용
        print(f"{idx}가 들어갔어요~~~")
        return self.data[idx]

    def __len__(self):
        print("코딩 좋아 ㅎㅎ")
        return len(self.data)

    def __str__(self):
        print("Hello World")
        return "매직 매소드"
```
