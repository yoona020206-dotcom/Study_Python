# 접근 지정자
- public
    - 외부에서 접속 가능
- private
    - 내부에서만 접속 가능

```python
class Student:

    def __init__(self, name, age):
        self.name = name # Public  Attribute
        self.age = age # Public  Attribute

    def print_msg(self): # Public  Method
        print("Hello World")
```

```python
std = Student("홍길동",33)
```

```python
std.name #외부에서 접속 혹은 호출
```

```python
std.print_msg() # 외부에서 호출
```

```python
class Student:

    def __init__(self, name, age):
        self.__name = name # Private Attribute 
        self.age = age # Public Attribute

    def __print_msg(self): # Private Method
        return f"{self.__name}님, {self.age}나이 축하드립니다." # Private Attribute &  Public Attribute

    def get_msg(self): # Public Method
        return self.__print_msg() # Private Method
```

```python
std = Student("홍길동",33)
```

```python
std.__name
```

```python
std.__print_msg()
```

```python
std.age
```

```python
std.get_msg()
```

# Decorator

```python
class student:
    def __init__(self, name):
        print("생성")
        self.__name = name

    @property
    # self.name = self.name()
    def name(self): # private 변수를 호출하는 함수 
        print("조회")
        return self.__name

    @name.setter
    def name(self, new_name):
        print("수정")
        self.__name = new_name

    @name.deleter
    def name(self):
        print("삭제")
        del self.__name
```

```python
def func():
    print("Hello World") # 여기는 실행 안됨

def func():
    print("코딩 좋아ㅎㅎ")

func()
```
