```python
# 변수 선언
a = "Hello World"
a
```

```python
a[0] # H 값이 리턴
```

```python
a[6]
```

```python
# 뒤로 셀때는 -붙힘
a[-5] # W 값이 리턴
```

```python
a[5], a[-6] # 공백 값이 리턴
```

```python
# 변수명 [시작인텍스(0), 끝인텍스(1)] : 문자열 슬라이싱
# a[0] + a[1] 둘다 같은 값
a[0:2] # 0 부터 1까지 리턴
```

```python
a[1:3]
```

```python
# 변수명 [시작인텍스(0), 끝인텍스(x): 간격] : 문자열 슬라이싱
# 첫번째 -> a[0] > H
# 두번째 -> a[0+1] > e
# 세번째 -> a[0+2] > 끝 인덱스와 숫자가 같아서 제외
a[0:2:1] # 0 부터 1까지 1칸씩 건너뛰며 리턴
```

```python
# 첫번째 -> a[0] > l
# 두번째 -> a[0+1] > l
# 세번째 -> a[0+2] > 끝 인덱스와 숫자가 같아서 제외
# .... 언제까지 해야할까? -11이 될때 까지..   
a[-9:-11:1]
```

```python
a[-9:-11:-1] # -를 붙혀서 반대로 만나야됨
```

```python
a[-1:-6:-2]
```

```python
a[0:3] = a[0:3:1]
```

```python
# 앞에 숫자가 0이면 생략 가능 (앞과 뒤 인덱스 생략가능) 뒤는 1일 때 

a[:3]
```

```python
a[0:]
```

```python
a[0:len(a):1]
```

```python
a[::]
```

```python
print("I eat %d apples." % 3) # I eat 3 apples.
```

```python
a = 5

type(a) # 꼭 숫자/문자인지 확인 할때 사용
print("I eat %d apples." % a) # I eat 5 apples.
```

```python
print("I eat %s apples." % "five") # I eat 5 apples.
```

```python
# 실무용
a = 5 

"{} eat {} apples.".format("I",a)
```

```python
a = 5 

"{1} eat {0} apples.".format("I",a) # 서로 위치 바꾸기 {1}에 a들어감, {0}에 I
```
