# 조건문


```python
a = 5

if a > 3: # a> 3 => 5> 3 => Ture
    print('a는 3보다 크다.')
else: # 그외 a<= 3
    print("a 는 3 보다 작거나 같다")
```

```python
# 부정을 반대 합니다. -> 긍정
not False
```

```python
if not a > 3: # not a> 3 => not 5> 3 => not Ture => False
    print('a는 3보다 크다.')
else: # 그외 a<= 3
    print("a 는 3 보다 작거나 같다")
```

```python
a = 5

if a > 3: # a가 3 보다 크면
    print("a는 3보다 크다.")
elif a > 1: # a가 3 보다 작거나 같으면서, 또는 1보다 크면
    print("a는 1 보다 크다.")
else: # a는 1 보다 작거나 같으면 
    print("a 는 3 보다 작거나 같다")
```
