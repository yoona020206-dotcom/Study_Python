````python
# 함수 선언
# 기본값(Default): p_x= 0
# -> 만약 입력하지 않았을 떄, 이때 는 기본 값을 적용
# -> 만약 입력하였을때, 이때는 기본값이 아닌 입력 값을 적용
def func(p_x=0, p_y=0): 
    return p_x + 10 + (p_y*2)

# 함수 호출
x = func(p_y =20)# p_y 값이 20 이고 싶으면 직접 입력 파이썬은 왼쪽부터 오른쪽으로 실행
x
````

### 가변 파라미터

````python
def add(p_x, p_y):
    return p_x + p_y


add(1,2)
````

````python
def add(*p_x): # *은 가변 파라미터를 뜻함
    add_sum = 0
    for i in p_x:
        add_sum = add_sum +i
    return add_sum


add(1,2,3,4,5,6,7,8)
````

````python
def add(*args): # *은 가변 파라미터를 뜻함, args를 변수명으로 많이 사용
    add_sum = 0
    print(type(args))
    for i in args:
        add_sum = add_sum +i
    return add_sum

add(1,2,3,4,5,6,7,8)
````

### 키워드 가변 파라미터

````python
def info(**kwargs): 
    print(type(kwargs))

info(p_name="홍길동", p_age=33)
````

````python
def info(**kwargs): 
    print(type(kwargs))
    print(f"{kwargs['p_name']}님은 나이가 {kwargs['p_age']}입니다.")

info(p_name="홍길동", p_age=33)
````

````python
def info(**kwargs): 
    print(type(kwargs))
    name = "신사임당"
    age = 18
    if 'p_name' in kwargs:
        name = kwargs['p_name']
    if 'p_age' in kwargs: # 여기서 elif를 쓰면 치명적인 오류 발생. if문이 참이면 elif를 수행하지 않기때문에 영원히 18이 돼버림.
        age = kwargs['p_age']

    print(f"{name}님은 나이가 {age}입니다.")

info(p_name="홍길동")
````
