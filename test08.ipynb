# 예외처리

```python
x  = "11"

int(x) # 문자를 숫자로 변환
```

```python
x = "홍길동"

try:
    print(f"{int(x)}는 숫자 입니다.")
except ValueError as e:
    print("오류에요")
```

```python
x = "홍길동"

try:
    print(f"{int(x)}는 숫자 입니다.")
except ValueError as e:
    print("오류에요")
finally:
    print("오류가 발생하든 안하든 무조건 실행하는 것")
```

```python
x = "ㅁㅁㅁ"

try:
    print(f"{int(x)}는 숫자 입니다.")
except ValueError as e:
    print("오류에요")
finally:
    print("오류가 발생하든 안하든 무조건 실행하는 것")
```

```python
x = "111"

try:
    print(f"{int(x)}는 숫자 입니다.")
except ValueError as e:
    print("오류에요")
else: # 아무런 문제 없이 실행 될때 나옴
    print("나 왔어~")
finally:
    print("오류가 발생하든 안하든 무조건 실행하는 것")
```

# assert 
- 특정 조건이 False일 때, 발생하는 에러

```python
x = 4

assert x % 3 == 0, "오류입니다"

print("오류가 아닙니다")
```

```python
x = 4

if x % 3 != 0:
    raise Exception("오류입니다")

print("오류가 아닙니다")
```
