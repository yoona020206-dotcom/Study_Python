# 리스트
- 문자열 & 숫자 -> 하나의 데이터 저장
- 리스트 -> 여러개의 데이터의 저장

```python
# 문자열
#""

# 숫자
# 3.14
# 모든 데이터는 변수 또는 함수에 담에서 씀

a = [ "Hello world", 0, 3.14 ]
```

```python
a
```

```python
len(a) # 문자의 개수를 쓸수 있을 때 씀
```

```python
len(0) # 숫자는 길이가 없어서 없음 (숫자는 크기가 큰거)
```

```python
# 길이가 존재하는 모든 데이터는 인덱스가 존재

a[0]
```

```python
a[-2]
```

```python
a[-1:-3:-1]
```

```python
# a[0][-1:-3:-1]  
# "Hello world"[-1:-3:-1]
a[0][-1:-3:-1]
```
