### collections

```python
from collections import Counter
```

```python
lst = [
    "aa", "aa", "aa", "bb", "bb", "bb", "bb", "tt"
]
```

```python
# 인스턴스 화
# counter: 객체 
counter = Counter(lst)
```

```python
counter
```

```python
type(counter)
```

```python
dict(counter)
```

```python
dict(counter)['aa']
```

```python
# 객체 => 변수 & 함수 
counter.elements()
```

```python
list(counter.elements())
```
