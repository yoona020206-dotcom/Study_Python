# 반복문
- 어떤 연산(기능)을 반복적으로 처리할때 사용

```python
print("2  * 1 = 2")
print("2  * 2 = 4")
print("2  * 3 = 6")
print("2  * 4 = 8")
print("2  * 5 = 10")
```

```python
x = 3 

print(f"{x}  * 1 = {x * 1}")
print(f"{x}  * 2 = {x * 2}")
print(f"{x}  * 3 = {x * 3}")
print(f"{x}  * 4 = {x * 4}")
print(f"{x}  * 5 = {x * 5}")
```

```python
x = 3

# x라는 변수의 값은 고정하지만
# i라는 변수는 print()가 실행될때 마다 변경됨.
for i in [1, 2, 3, 4, 5]:
    print(f"{x}  * i = {x * i}")

"""
for문 실행시,,,
1번째 실행 i = [1, 2, 3, 4, 5][0] & x = 3
2번째 실행 i = [1, 2, 3, 4, 5][1] & x = 3
3번째 실행 i = [1, 2, 3, 4, 5][2] & x = 3 ...
"""
```

```python
3 % 2 # 나머지 값 리턴
```

```python
3 % 2 == 0 # 만약 참이면, 짝수 
```

```python
for i in [1, 2, 3, 4, 5]:
    if i % 2 == 0:
        print(f"{x}  * i = {x * i}")
```

```python
for i in [1, 2, 3, 4, 5]:
    if i % 2 == 0:
        pass # 아무런 행위를 하지 않음
        print(f"{x}  * i = {x * i}")
```

```python
for i in [1, 2, 3, 4, 5]:
    if i % 2 == 0:
        print(f"{x}  * i = {x * i}")
        break # 처음만 실행 나머지 실행 안됨 (위에서 아래로 좌에서 우로 연산함)
```

```python
"""
1번째 실행 했을때 i = 1 > 거짓> print()실행 안됨
2번째 실행 했을때 i = 2 > 참> break > print()실행 안됨
3 번째 실행 없음
""" 

for i in [1, 2, 3, 4, 5]:
    if i % 2 == 0:#
        break # 뒤 부터 나머지 실행 안됨 
        print(f"{x}  * i = {x * i}")
```

```python
"""
1번째 실행 했을때 i = 1 > 거짓> print()실행 안됨
2번째 실행 했을때 i = 2 > 참> countinue > print()실행 안됨
3 번째 실행 했을때 i = 3 > 거짓> print()실행 안됨
4번째 실행 했을때 i = 4 > 참> countinue > print()실행 안됨
5 번째 실행 했을때 i = 5 > 거짓> print()실행 안됨
""" 

for i in [1, 2, 3, 4, 5]:
    if i % 2 == 0:
        continue # 다시 for로 돌아감
        print(f"{x}  * i = {x * i}")
```

```python
for i in [1, 2, 3, 4, 5]:
    if i % 2 == 0:
        continue # 다시 for로 돌아감
        print(f"{x}  * i = {x * i}")
else: 
    print("for문이 정상적으로 종료됨")
```
